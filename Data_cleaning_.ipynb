```python
q1 = df['Fare'].quantile(0.25)
q3 = df['Fare'].quantile(0.75)

Iqr = q3 - q1

upper = q3 + 1.5 * Iqr

print('q1:', q1, 'q3:', q3, 'upper limit:', round(upper, 2))
print('outliers:', (df['Fare'] > upper).sum())
```

```python
df.loc[df['Fare'] > upper, 'Fare'] = upper
df['Fare'].describe()
```
